# HW1 – TuriCreate SFrame Practice

This notebook loads the **Wiki People SFrame** and answers the HW1 questions:

1. Load the SFrame.
2. Count the number of rows.
3. Find the name in the last row.
4. Read the text for **Harpdog Brown**.
5. Sort by the `text` column and find the first name.


```python
import zipfile
import os

zip_path = "wiki_people.zip"

with zipfile.ZipFile(zip_path, 'r') as z:
    z.extractall(".")

print("Unzipped files:")
os.listdir(".")
```

```python
import turicreate as tc

sf = tc.SFrame("people_wiki.sframe")
sf
```

## Question 2 – How many rows are in the SFrame?

```python
num_rows = sf.num_rows()
print('Number of rows in the SFrame:', num_rows)
num_rows
```

## Question 3 – Which name is in the last row?

```python
last_name = sf[-1]['name']
print('Name in the last row:', last_name)
last_name
```

## Question 4 – Read the text column for Harpdog Brown

```python
harpdog = sf[sf['name'] == 'Harpdog Brown']
harpdog_text = harpdog['text'][0]
harpdog_text
```

## Question 5 – Sort by the `text` column (ascending) and find the first name

```python
sf_sorted = sf.sort('text', ascending=True)
first_name = sf_sorted[0]['name']
print('First name after sorting by text (ascending):', first_name)
first_name
```
